```python
print("==Travel Agent==")
print("1. Lahore")
print("2. Karachi")
print("3. Islamabad")

choice = int(input( "Enter your destination (1-3): "))

if choice == 1:
    print("You selected Lahore")
    print("Ticket Price: Rs. 5000")
elif choice == 2:
    print("You selected Karachi")
    print("Ticket Price: Rs. 8000")

elif choice == 3:
    print("You selected Islamabad")
    print("Ticket Price: Rs. 6000")

else:
    print("Invalid choice")

print("Thank you for using our Travel Agent!")

```
